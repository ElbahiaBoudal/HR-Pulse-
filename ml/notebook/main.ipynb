```python
df = df.drop(columns=["index"])
```

```python
df['Rating'].mean()
```

```python
df['Rating'] = df['Rating'].replace(-1, np.nan).fillna(df['Rating'].mean())
df['Founded'] = df['Founded'].replace(-1, np.nan).fillna(df['Founded'].median())
```

```python
df['Rating'].mean()
```

```python
df['Competitors']=df['Competitors'].replace("-1", "Unknown", inplace=True)
```

```python
df["Company Name"] = df["Company Name"].str.split("\n").str[0]
```

```python
df["State"] = df["Location"].str.split(",").str[1]
df["Location"] = df["Location"].str.split(",").str[0]
```

```python
df["Size"] = df["Size"].replace("-1", "Unknown")
```

```python
import re
def clean_salary(salary):
    nums = re.findall(r"\d+", salary)
    if len(nums) >= 2:
        return (int(nums[0]) + int(nums[1])) / 2
    return None

df["salary_avg"] = df["Salary Estimate"].apply(clean_salary)
df[["Salary Estimate", "salary_avg"]].head()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['salary_avg'], kde=True, color='blue')
plt.title('Distribution du Salaire Moyen')
plt.xlabel('Salaire Annuel ($K)')
plt.ylabel('Fréquence')
plt.show()
```

```python
import string
def normalize_job_title(title):
    title = str(title).lower().strip()  # lowercase + trim
    title = title.translate(str.maketrans("", "", string.punctuation))  # remove punctuation
    return title

df["Job Title"] = df["Job Title"].apply(normalize_job_title)
```

```python

import nltk
from nltk.corpus import stopwords

# Download stopwords (une seule fois)
nltk.download("stopwords")

# Stopwords English
stop_words = set(stopwords.words("english"))

# 🔹 Fonction complète de nettoyage
def clean_job_description(text):
    if pd.isna(text):
        return ""

    # lowercase
    text = text.lower()

    # remove new lines
    text = re.sub(r"\n", " ", text)

    # remove numbers
    text = re.sub(r"\d+", "", text)

    # remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # remove stopwords
    words = text.split()
    words = [w for w in words if w not in stop_words]

    return " ".join(words)

# Apply cleaning
df["Job Description"] = df["Job Description"].apply(clean_job_description)
```
